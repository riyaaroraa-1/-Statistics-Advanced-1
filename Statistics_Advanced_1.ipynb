**Q1: What is a random variable in probability theory?**

Ans. A random variable in probability theory is a variable that assigns numerical values to the outcomes of a random experiment. It is used to represent the outcomes of uncertain events mathematically.

For example, when a die is rolled, the random variable can take value from 1 to 6 depending on the outcome.


**Q2: What are the types of random variables?**

Ans. The two main types of random variables are:

1. Discrete Random Variable
   A discrete random variable takes countable values, such as whole numbers.
   Example: Number of students present in a class

  2. Continuous Random Variable
     A continuous random variable can take any value within a given range.
     Example: Height or weight of a person.

     

**Q3: Explain the difference between discrete and continuous distributions.**

Ans. A discrete distribution and a continuous distribution differ mainly in the type of values their random variables can take.

Discrete Distribution

A discrete distribution describes the probability of outcomes that are countable and distinct. The values usually occur in whole numbers and there are no intermediate values between them.

Examples:

* Number of students in a classroom
* Number of cars sold in a day
* Number obtained when rolling a die

In a discrete distribution, the probability of each exact value can be calculated individually.

**Continuous Distribution**

A continuous distribution describes the probability of outcomes that can take any value within a given range. These values are measurable and can include decimals or fractions.

Examples:
* Height and weight of individuals
* Temperature readings
* Time taken to complete a task

In a continuous distribution, probabilities are calculated over interval rather than at exact points because there are infinitely many possible values.


**Q4: What is a binomial distribution, and how is it used in probability?**

Ans. A binomial distribution is a type of discrete probability distribution that shows the probability of obtaining a fixed number of successful outcomes in a specific number of independent trials. Each trial has only two possible outcomes: success or failure.

A binomial distribution is based on the following conditions:

The number of trial is fixed.
Each trial is independent of others.
There are only two possible outcomes for each trial.
The probability of success remains constant in every trial

Example:

If a coin is tossed 10 times, the binomial distribution can be used to calculate the probability of getting exactly 6 heads. In this case:

Number of trials(n) = 10
Probability of success(p)= 0.5
Success = getting a head
Uses of Binomial Distribution

The binomial distribution is widely used in probability and statistics to analyze situations involving repeated experiments with two possible outcomes. It is commonly used in:

* Quality control to determine defective products
* Survey analysis for yes/no responses
* Medical studies to measure treatment success
* Business and marketing research to predict customer behavior.

Thus the binomial distribution helps in calculating and predicting the likelihood of different outcomes in real-life situations involving uncertainty.


**Q5: What is the standard normal distribution, and why is it important?**

Ans. The standard normal distribution is a special type of normal distribution in which the mean is equal to 0 and standard deviation is equal to 1. It is represented by a bell-shaped, symmetrical curve where most values are concentrated around the mean and the probabilities decrease as we move away from the center.

In a standard normal distribution:
* Mean (μ) = 0
* Standard Deviation (σ) = 1

The values in this distribution are expressed as z-scores, which indicate how far a data value is from the mean in terms of standard deviations.

Importance of Standard Normal Distribution

The standard normal distribution is important because it provides a common standard for comparing different datasets and calculating probabilities. It is widely used in statistics and probability theory for data analysis and decision-making.

Applications
* Calculating probabilities using z - tables
* Standardizing different data values for comparison
* Performing hypothesis testing
* Conducting statistical analysis in research and business
* Measuring confidence intervals and error margins

Example:
If students exam scores follow a normal distribution, the standard normal distribution can help determine how a particular student performed compared to the average class performance using z - scores.

Therefore, the standard normal distribution plays a vital role in statistical analysis by simplifying calculations and helping researchers interpret data accurately.

**Q6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?**

Ans. The Central Limit Theorem (CLT) is an important concept in statistics. It states that when a sufficiently large random sample is taken from a population, the distribution of a sample mean becomes approximately normal, regardless of the original distribution of the population.

In simple terms, even if the original data is not normally distributed, the averages of large samples will tend to follow a normal distribution.

The Central Limit Theorem is critical in statistics because it forms the basis for many statistical methods and analyses. It helps researchers and analysts make predictions and draw conclusions about a population using sample data.

The CLT is widely used in:

* Hypothesis testing
* Confidence interval estimation
* Data analysis and forecasting
* Business and scientific research

Therefore, the Central Limit Theorem plays a vital role in statistical analysis by simplifying complex data and enabling accurate decision-making.


**Q7: What is the significance of confidence intervals in statistical analysis?**

Ans. Confidence intervals are an important concept in statistical analysis because they provide a range of values within which the true population parameter is expected to lie with a certain level of confidence.

A confidence interval is calculated using sample data and helps measure the reliability and accuracy of an estimate. Instead of giving a single value, it provides an interval that is more informative and realistic.

For example, a 95% confidence interval means that there is a 95% confidence that the true population mean lies within the calculated range.

Confidence intervals are significant because they:

* Help estimate population parameters accurately

* Measure the uncertainty in sample estimates

* Support decision-making and predictions

* Are widely used in hypothesis testing and research analysis

* Improve the reliability of statistical conclusions

Thus, confidence intervals are essential in statistics as they help analysts and researchers make informed decisions based on sample data.


**Q8: What is the concept of expected value in a probability distribution?**

Ans. The expected value in a probability distribution is the long-run average value of a random variable after many repeated experiments. It represents the value we “expect” to occur on average.

For a discrete probability distribution, the expected value is calculated by multiplying each possible outcome by its probability and then adding all the results together.

E(X)=∑x⋅P(x)

Where:

* x = possible outcomes

* P(x) = probability of each outcome

Expected value is important in statistics, economics, finance, and decision-making because it helps predict average outcomes and analyze risk and uncertainty.



**Q9: Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
deviation using NumPy, and draw a histogram to visualize the distribution.**

````python
import numpy as np
import matplotlib.pyplot as plt

# Generating 1000 random numbers from a normal distribution

data = np.random.normal(loc=50, scale=5, size=1000)

# Compute mean and standard deviation
mean_val = np.mean(data)
std_dev = np.std(data)

print("Mean:", mean_val)
print("Standard Deviation:", std_dev)

# Histogram
plt.hist(data, bins=30, edgecolor = 'black')
plt.title("Histogram of Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
````

**Question 10: You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.**
```
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
```
* **Explain how you would apply the Central Limit Theorem to estimate the average sales
with a 95% confidence interval.**
* **Write the Python code to compute the mean sales and its confidence interval.**


Ans. The Central Limit Theorem (CLT) states that for a sufficiently large sample size, the sampling distribution of the sample mean becomes approximately normally distributed, regardless of the original population distribution.

In this problem, the daily sales data of a retail company is analyzed to estimate the average sales along with the 95% confidence interval. The sample mean represents the average daily sales, while the confidence interval provides a range within which the true population mean is likely to lie.

First, the sample mean and the sample standard deviation are calculated from the sales data. Then, the standard error is computed.

where:
* (s) is the sample standard deviation
* (n) is the sample size

For a 95% confidence interval, the z-value of 1.96 is used. The confidence interval is calculated.

This interval helps the company estimate the range of average daily sales with 95% confidence and supports better business decision-making.

````python
import numpy as np
import scipy.stats as st

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
235, 260, 245, 250, 225, 270, 265, 255, 250, 260]


mean_sales = np.mean(daily_sales)
std_sales = np.std(daily_sales, ddof = 1)

# sample size
n = len(daily_sales)

# Confidence interval (95%)
confidence_level = 0.95
alpha = 1 - confidence_level
z_score = st.norm.ppf(1 - alpha/2)

margin_of_error = z_score * (std_sales / np.sqrt(n))

lower_bound = mean_sales - margin_of_error
upper_bound = mean_sales + margin_of_error

print(f"Mean Sales: {mean_sales:.2f}")
print(f"95% Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")

````
